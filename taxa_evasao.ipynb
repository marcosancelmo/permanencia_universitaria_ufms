A taxa de evasão pode ser calculada a partir da quantidade de alunos ingressos e egressos, considerando a duração dos cursos. A fórmula básica é: Taxa de Evasão = (1 − (Número de Egressos / Número de Ingressos)) × 100

```python
##Inicialmente preferi realizar o trabalho manual de importar os databases para o python mas vou evoluir o projeto para implemnetar automoção por API

##Egressos
e21 = 'https://dadosabertos.ufms.br/dataset/9eda7d71-bb81-46b4-8e3c-9161346b6ebc/resource/733b0bc8-c616-41e1-93b5-b912ff4038a9/download/siscad-lista-egressos-2021.csv'
e22 = 'https://dadosabertos.ufms.br/dataset/9eda7d71-bb81-46b4-8e3c-9161346b6ebc/resource/9aa9035a-cdd9-44e0-b8f1-cec1b692db5d/download/lista_de_egresso_2022.csv'
e23 = 'https://dadosabertos.ufms.br/dataset/9eda7d71-bb81-46b4-8e3c-9161346b6ebc/resource/f0f2d176-a5b3-483f-a535-a1ed32bd9f96/download/siscad-lista-egressos-2023.csv'
e24 = 'https://dadosabertos.ufms.br/dataset/9eda7d71-bb81-46b4-8e3c-9161346b6ebc/resource/ad513f7b-9a0b-4130-836b-726260b7ba6b/download/siscad-lista-egressos-2024.csv'
e25 = 'https://dadosabertos.ufms.br/dataset/9eda7d71-bb81-46b4-8e3c-9161346b6ebc/resource/41cb7797-802c-4d7b-9502-5a8529651bcb/download/siscad-lista-egressos-2025.csv'

##Ingressos
i21 = 'https://dadosabertos.ufms.br/dataset/c08666c6-01a5-488a-9ff4-d1539ee33d7d/resource/18a93b16-c49c-4107-9dae-fb4b8c6e4c59/download/siscad-qtd-ingressos-2021.csv'
i22 = 'https://dadosabertos.ufms.br/dataset/c08666c6-01a5-488a-9ff4-d1539ee33d7d/resource/37b68d73-4bba-4684-b27f-a624cff17d66/download/quantidade_ingresso_graduacao_2022.csv'
i23 = 'https://dadosabertos.ufms.br/dataset/c08666c6-01a5-488a-9ff4-d1539ee33d7d/resource/3eb22a3c-cc9b-4f2e-8aba-059a569b20da/download/siscad-qtd-ingressos-2023.csv'
i24 = 'https://dadosabertos.ufms.br/dataset/c08666c6-01a5-488a-9ff4-d1539ee33d7d/resource/808f08d2-0b15-4109-97e6-9eb7f8169948/download/siscad-qtd-ingressos-2024.csv'
i25 = 'https://dadosabertos.ufms.br/dataset/c08666c6-01a5-488a-9ff4-d1539ee33d7d/resource/59b5b6f1-fba7-43e8-8bfc-636bb4785f06/download/siscad-qtd-ingressos-2025.csv'
basei25 = pd.read_csv(i25)
basei24 = pd.read_csv(i24)
basei23 = pd.read_csv(i23)
basei22 = pd.read_csv(i22)
basei21 = pd.read_csv(i21)

ingres_tot = pd.concat([basei25,basei24,basei23,basei22,basei21], ignore_index=True)
                      
basee25 = pd.read_csv(e25)
basee24 = pd.read_csv(e24)
basee23 = pd.read_csv(e23)
basee22 = pd.read_csv(e22)
basee21 = pd.read_csv(e21)

egres_tot = pd.concat([basee25,basee24,basee23,basee22,basee21], ignore_index=True)

ingres_tot.head()
```

```python
## idxmin e idxmax procura o id do minimo e do náximo
id_max = base25['qtd_ingressos'].idxmax()
print(base25.loc[[id_max]])
```

```python
## Aqui realizei a padronização da tabea de Egressos para poder juntar futuramente com a de Ingressos e manipular o cálculo da evasão
egress_trat = egres_tot.groupby(['ano', 'semestre', 'unidade', 'curso',])['nome'].count().reset_index(name='qtd_egressos')
print(egress_trat.head(10))
egress_trat
```

```python
# 1. Faz o merge trazendo apenas a chave de junção e a nova coluna desejada
df_geral = pd.merge(
    ingres_tot,
    egress_trat[['ano', 'semestre', 'unidade', 'curso', 'qtd_egressos']], # Seleciona só a chave + a coluna nova do df2 que é qtd_egressos
    on= ['ano', 'semestre', 'unidade', 'curso'],              # Coluna que serve de chave em ambos
    how='left'                                                # Mantém todas as linhas de df1
)
```

```python
df_geral
```

```python
##(1 − (Número de Egressos / Número de Ingressos)) × 100
df_geral['evasao'] = (1 - (df_geral['qtd_egressos']  / df_geral['qtd_ingressos'])) * 100
df_geral['evasao'] = df_geral['evasao'].round(2) ## round(1) arredondando as casas decimais
```

```python
##Sem evasão
sem_evasao = df_geral[df_geral.isnull().any(axis=1)]
sem_evasao
```

```python
## Apenas com evasao
com_evasao = pd.concat([df_geral, sem_evasao, sem_evasao]).drop_duplicates(keep=False)
com_evasao['qtd_egressos'] = com_evasao['qtd_egressos'].astype(np.int64)
com_evasao
```
